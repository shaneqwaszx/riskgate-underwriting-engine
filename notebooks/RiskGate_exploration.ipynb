# 🏦 Lending Club

# 📝 Business Understanding

> - If the applicant is likely to repay the loan, then not approving the loan results in a loss of business to the company
> - If the applicant is not likely to repay the loan, i.e. he/she is likely to default, then approving the loan may lead to a financial loss for the company

> The data given contains the information about past loan applicants and whether they ‘defaulted’ or not. The aim is to identify patterns which indicate if a person is likely to default, which may be used for takin actions such as denying the loan, reducing the amount of loan, lending (to risky applicants) at a higher interest rate, etc.

> When a person applies for a loan, there are two types of decisions that could be taken by the company:
> 1. `Loan accepted`: If the company approves the loan, there are 3 possible scenarios described below:
    - `Fully paid`: Applicant has fully paid the loan (the principal and the interest rate)
    - `Current`: Applicant is in the process of paying the instalments, i.e. the tenure of the loan is not yet completed. These candidates are not labelled as 'defaulted'.
    - `Charged-off`: Applicant has not paid the instalments in due time for a long period of time, i.e. he/she has defaulted on the loan
> 2. `Loan rejected`: The company had rejected the loan (because the candidate does not meet their requirements etc.). Since the loan was rejected, there is no transactional history of those applicants with the company and so this data is not available with the company (and thus in this dataset)

# 🎯 Business Objectives
> - `LendingClub` is the largest online loan marketplace, facilitating personal loans, business loans, and financing of medical procedures. Borrowers can easily access lower interest rate loans through a fast online interface. 
> - Like most other lending companies, lending loans to ‘`risky`’ applicants is the largest source of financial loss (called `credit loss`). The credit loss is the amount of money lost by the lender when the borrower refuses to pay or runs away with the money owed. In other words, borrowers who defaultcause the largest amount of loss to the lenders. In this case, the customers labelled as '`charged-off`' are the '`defaulters`'. 
> - If one is able to identify these risky loan applicants, then such loans can be reduced thereby cutting down the amount of credit loss. Identification of such applicants using EDA and machine learning is the aim of this case study. 
> - In other words, the company wants to understand the driving factors (or driver variables) behind loan default, i.e. the variables which are strong indicators of default. The company can utilise this knowledge for its portfolio and risk assessment.

## 1d. Relabelling the Loan Status

From the above visualisation, the two loan status we are interested in are <b>Defaulted</b> (Charged Off) and <b>Not Defaulted</b> (Fully Paid). Hence, we will need to add a new variable which will be binary (0s and 1s).

- 0 means Not Defaulted
- 1 means Defaulted

All those loans, whose status is “Fully Paid”, “Current” will be categorized as Not Defaulted and anything else will be categorized as Defaulted. To achieve this we will introduce new variable defaulted.

```python
# First we define the function
def change_loan_status(loan_status):
    if loan_status == 'Fully Paid':
        return 0
    else:
        return 1

# Next we apply the function
train_df['loan_status'] = train_df['loan_status'].apply(change_loan_status)
train_df.head()
```

# 3.🔄 Feature Engineering
---

> **OVERALL GOAL:**

First we convert the text data into useful categorical numerical variable

```python
# Scaling numerical columns using MinMaxScaler()
from sklearn import preprocessing
import pandas as pd
scaled_train_df = train_df.copy()
scaler = preprocessing.MinMaxScaler()
processing_train_df = scaled_train_df.select_dtypes(include = ['float64', 'int64'])
names = processing_train_df.columns
d = scaler.fit_transform(processing_train_df)
scaled_train_df = pd.DataFrame(d, columns=names)
str_train_df = train_df.copy().select_dtypes(include = 'object')
new_scaled_train_df = pd.concat([scaled_train_df,str_train_df], axis="columns")
new_scaled_train_df['zip_code'] = new_scaled_train_df['zip_code'].astype(int)
new_scaled_train_df.info()
```

```python
# categorical tranformation 2
# Name it cate_train_df to represent categorical transformation done on this dataframe
cate_train_df = new_scaled_train_df.copy()

# Ordinal Transformation
cate_train_df['term'] = cate_train_df['term'].replace({' 60 months':60,' 36 months':36})
cate_train_df['initial_list_status'] = cate_train_df['initial_list_status'].replace({'f':0,'w':1})
cate_train_df['emp_length'] = cate_train_df['emp_length'].replace({'10+ years':10,'0':0,'7 years':7,
                                                                     '9 years':9,'5 years':5,'1 year':1,
                                                                     '2 years':2,'3 years':3,'6 years':6,
                                                                     '4 years':4,'8 years':8,'< 1 year':0.5})
cate_train_df['verification_status'] = cate_train_df['verification_status'].replace({'Verified':1,'Not Verified':0,'Source Verified':2})
cate_train_df['grade'] = cate_train_df['grade'].replace({'A':7,'B':6,'C':5,'D':4,'E':3,'F':2,'G':1})
cate_train_df['sub_grade'] = cate_train_df['sub_grade'].replace({'D3':4.3,'C2':5.2,'B1':6.1,'F1':2.1,'C3':5.3,'A3':7.3,'B5':6.5,
                                                       'B2':6.2,'D4':4.4,'B3':6.3,'F3':2.3,'D1':4.1,'A2':7.2,'C5':5.5,
                                                       'E1':3.1,'A4':7.4,'C1':5.1,'C4':5.4,'D2':4.2,'B4':6.4,'A5':7.5,
                                                       'E3':3.3,'A1':7.1,'D5':4.5,'F2':2.2,'E4':3.4,'E2':3.2,'G2':1.2,
                                                       'E5':3.5,'F5':2.5,'F4':2.4,'G3':1.3,'G5':1.5,'G1':1.1,'G4':1.4})
cate_train_df['application_type'] = cate_train_df['application_type'].replace({'INDIVIDUAL':1,'DIRECT_PAY':0,'JOINT':2})

# Freqeuncy Encoding for non-numerical colums with no ordinal relationship
# Purpose column
enc_purpose = (cate_train_df.groupby('purpose').size()) / len(cate_train_df)
cate_train_df['purpose'] = cate_train_df['purpose'].apply(lambda x : enc_purpose[x])
# Home ownership column
enc_home_ownership = (cate_train_df.groupby('home_ownership').size()) / len(cate_train_df)
cate_train_df['home_ownership'] = cate_train_df['home_ownership'].apply(lambda x : enc_home_ownership[x])
# Employment title
enc_emp_title = (cate_train_df.groupby('emp_title').size()) / len(cate_train_df)
cate_train_df['emp_title'] = cate_train_df['emp_title'].apply(lambda x : enc_emp_title[x])
# Title
enc_title = (cate_train_df.groupby('title').size()) / len(cate_train_df)
cate_train_df['title'] = cate_train_df['title'].apply(lambda x : enc_title[x])
# State
enc_state = (cate_train_df.groupby('state').size()) / len(cate_train_df)
cate_train_df['state'] = cate_train_df['state'].apply(lambda x : enc_state[x])

cate_train_df.info()
```

```python
# Feature importance selection using XGBoost to generate the best possible variables for input feature.

import xgboost
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder

x_features = ['loan_amnt','int_rate','installment','annual_inc','dti','open_acc','pub_rec','revol_bal','revol_util',
              'total_acc', 'mort_acc','pub_rec_bankruptcies','term','grade','sub_grade','emp_title', 
              'emp_length','home_ownership','verification_status','purpose','title','initial_list_status',
              'application_type','zip_code','state']
X_boost = cate_train_df.loc[:,x_features]
y_boost = cate_train_df.loc[:,'loan_status']
 
le = LabelEncoder()
y = le.fit_transform(y_boost)

model = xgboost.XGBClassifier(n_estimators = 1000, learning_rate = 0.3, max_depth = 5)
model.fit(X_boost,y_boost)

feat_imp = pd.Series(model.feature_importances_, index=X_boost.columns)
feat_imp.nlargest(14).plot(kind='barh')

plt.figure(figsize=(8,6))
plt.show()
```

```python
def simulate_policy(p, y_true, t_low, t_high, lgd=1.0, ead=1.0):
    """
    Policy:
    - p < t_low  => auto-approve
    - t_low<=p<t_high => manual review
    - p >= t_high => reject

    Expected loss proxy (approve + review accepted cases):
    For proposal, assume review cases are still approved (worst-case) OR
    you can assume review reduces PD by a factor later.
    """
    bucket = np.where(p < t_low, "approve", np.where(p < t_high, "review", "reject"))
    approve_mask = bucket != "reject"

    exp_loss = (p[approve_mask] * lgd * ead).sum()
    default_rate_approved = y_true[approve_mask].mean()
    review_rate = (bucket == "review").mean()
    approve_rate = (bucket == "approve").mean()
    reject_rate = (bucket == "reject").mean()

    return {
        "t_low": t_low, "t_high": t_high,
        "approve_rate": approve_rate,
        "review_rate": review_rate,
        "reject_rate": reject_rate,
        "default_rate_nonrejected": default_rate_approved,
        "expected_loss_proxy": exp_loss
    }

# Sweep t_high while holding t_low fixed (illustrative)
rows = []
t_low_fixed = 0.10
for t_high in np.linspace(0.15, 0.60, 15):
    rows.append(simulate_policy(cal_proba, y_test, t_low_fixed, float(t_high)))

sim = pd.DataFrame(rows)

plt.figure()
plt.plot(sim["t_high"], sim["expected_loss_proxy"])
plt.title("Risk Exposure Proxy (ΣPD) vs Reject Threshold (t_high)")
plt.xlabel("t_high")
plt.ylabel("Risk Exposure Proxy (ΣPD)")
plt.tight_layout()
plt.savefig("fig_risk_exposure_vs_threshold.png", dpi=200)

plt.figure()
plt.plot(sim["t_high"], sim["review_rate"], label="Review rate")
plt.plot(sim["t_high"], sim["reject_rate"], label="Reject rate")
plt.title("Operational Load vs Reject Threshold (t_high)")
plt.xlabel("t_high")
plt.ylabel("Rate")
plt.legend()
plt.tight_layout()
plt.savefig("fig_ops_load_vs_threshold.png", dpi=200)

print("Saved: fig_expected_loss_vs_threshold.png, fig_ops_load_vs_threshold.png")
sim.head()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import precision_score, recall_score, f1_score

p = cal_proba if "cal_proba" in globals() else xgb_proba6

# Use the chosen threshold (from your previous selection)
chosen_t = 0.35

# Still compute the curve for context
thresholds = np.round(np.arange(0.05, 1.00, 0.05), 2)

rows = []
for t in thresholds:
    y_pred = (p >= t).astype(int)
    rows.append({
        "threshold": t,
        "precision": precision_score(y_test, y_pred, zero_division=0),
        "recall": recall_score(y_test, y_pred, zero_division=0),
        "f1": f1_score(y_test, y_pred, zero_division=0)
    })

tm = pd.DataFrame(rows)

# Metrics at the chosen threshold (0.35)
row_035 = tm.loc[tm["threshold"] == chosen_t].iloc[0]
prec_035 = row_035["precision"]
rec_035  = row_035["recall"]
f1_035   = row_035["f1"]

plt.figure(figsize=(5.2, 3))
plt.plot(tm["threshold"], tm["precision"], label="Precision")
plt.plot(tm["threshold"], tm["recall"], label="Recall")
plt.plot(tm["threshold"], tm["f1"], label="F1")

# Highlight threshold = 0.35
plt.axvline(chosen_t, linestyle="--", label=f"Chosen t={chosen_t:.2f}")
plt.scatter([chosen_t], [prec_035], s=40, label=f"Prec@0.35={prec_035:.2f}")
plt.scatter([chosen_t], [rec_035],  s=40, label=f"Rec@0.35={rec_035:.2f}")
plt.scatter([chosen_t], [f1_035],   s=40, label=f"F1@0.35={f1_035:.2f}")

plt.xlabel("Threshold")
plt.ylabel("Score")
plt.title("Threshold Trade-off (Chosen threshold = 0.35)")
plt.legend(loc="best")
plt.tight_layout()
plt.savefig("fig_threshold_tradeoff_035.png", dpi=200, bbox_inches="tight")
plt.show()

print("Saved: fig_threshold_tradeoff_035.png")
print(f"At threshold 0.35 -> Precision={prec_035:.4f}, Recall={rec_035:.4f}, F1={f1_035:.4f}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

p = cal_proba if "cal_proba" in globals() else xgb_proba6
t_low = 0.10
cap = 0.20

rows = []
for t_high in np.linspace(0.15, 0.60, 15):
    bucket = np.where(p < t_low, "approve", np.where(p < t_high, "review", "reject"))
    rows.append({
        "t_high": float(t_high),
        "review_rate": np.mean(bucket == "review"),
        "reject_rate": np.mean(bucket == "reject")
    })

sim = pd.DataFrame(rows)

# choose largest t_high where review_rate <= cap
feasible = sim[sim["review_rate"] <= cap]
chosen = float(feasible["t_high"].max()) if len(feasible) else float(sim["t_high"].iloc[0])

plt.figure(figsize=(5,3))
plt.plot(sim["t_high"], sim["review_rate"], label="Review rate")
plt.plot(sim["t_high"], sim["reject_rate"], label="Reject rate")
plt.axhline(cap, linestyle="--", label=f"Capacity cap={cap:.0%}")
plt.axvline(chosen, linestyle="--", label=f"Chosen t_high={chosen:.2f}")
plt.xlabel("t_high")
plt.ylabel("Rate")
plt.title("Operational Load vs Reject Threshold (t_high)")
plt.legend()
plt.tight_layout()
plt.savefig("fig_ops_load_with_cap.png", dpi=200, bbox_inches="tight")
plt.show()
print("Saved: fig_ops_load_with_cap.png")
```
